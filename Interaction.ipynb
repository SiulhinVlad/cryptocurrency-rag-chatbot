Interaction with fine-tuned model and implementing Graphical User Interface

```python
import transformers
import torch
from transformers import AutoTokenizer
from ipywidgets import widgets, Layout
from IPython.display import display

model = "siulhin-vlad37/Llama-2-7b-chat-bot"

prompt_text = widgets.Text(
    placeholder="Enter prompt",
    description="Prompt:",
    layout=Layout(width="600px"),
)

output_text = widgets.Textarea(
    value="",
    placeholder="Generated Text",
    description="Output:",
    layout=Layout(width="600px", height="100px"),
)

generate_button = widgets.Button(description="Generate")

def generate_text(b):
    prompt = prompt_text.value

    tokenizer = AutoTokenizer.from_pretrained(model)
    pipeline = transformers.pipeline(
        "text-generation",
        model=model,
        torch_dtype=torch.float16,
        device_map="auto",
    )

    sequences = pipeline(
        f'[INST] {prompt} [/INST]',
        do_sample=True,
        top_k=10,
        num_return_sequences=1,
        eos_token_id=tokenizer.eos_token_id,
        max_length=300,
    )

    generated_text = sequences[0]['generated_text']
    generated_text = generated_text.replace(f'[INST] {prompt} [/INST]', '').replace('[INST]', '').replace('[/INST]', '')
    output_text.value = generated_text.strip()

generate_button.on_click(generate_text)

display(prompt_text)
display(generate_button)
display(output_text)
```
